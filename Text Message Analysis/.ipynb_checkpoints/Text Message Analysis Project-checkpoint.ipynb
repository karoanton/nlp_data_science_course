## Text preprocessing

### Data cleaning

To begin, I want to check for any null values and remove those from the dataframe.

```python
df.isna().sum()
```

The **Message** column contains three null values. Let's go ahead and remove them from the dataframe.

```python
df = df.dropna()
```

```python
df.isna().sum()
```

The **Date** column in the dataframe displays a string consisting of a year and month. I want to split this into two columns--**Year** and **Month**--and convert the values into integers.

```python
df[["Year", "Month"]] = df["Date"].str.split("/", expand=True).astype(int)
```

```python
df.head()
```

```python
df.info()
```

Now that the values of **Date** have been split and converted into integers, that column can be deleted from the dataframe.

```python
df = df.drop(columns=["Date"])
```

```python
df.head()
```

The column labelled **length** refers to the character length of each text message included in the column **Message**.

First, I'll go ahead and rename the column label as **Characters**. Then, I will convert the string values of the column into integers.

```python
df = df.rename(columns={"length": "Characters"})
```

```python
df["Characters"] = df["Characters"].astype(int)
```

When I call *.astype(int)* to convert the values of **Characters** from strings to integers, I get a ValueError message: "invalid literal for int() with base 10: 'Err:511'".

Let's filter for "Err" to see how many rows feature an error value.

```python
filtered_df = df[df['Characters'].str.contains("Err")]
print(filtered_df)
```

There four rows featuring error messages for the **Message** and **Characters** column. Let's remove these rows from the final dataframe.

```python
df = df.drop([9468, 18485, 31264, 32286])
```

```python
df.info()
```

Now that the "Err" values have been removed from the dataframe, I can try to convert the values in **Characters** to integers again.

```python
df["Characters"] = df["Characters"].astype(int)
```

```python
df.info()
```

Finally, I want to clean up the column **country**. Let's start by fixing this column label as well.

```python
df = df.rename(columns={'country': 'Country'})
```

```python
df.info()
```

Now let's go ahead and fix the values of that column. If you look back at **filtered_df**, you can see the values *SG* and *Singapore*, both of which refer to the nation Singapore.

I want to edit the country values to ensure all messages from a single country are properly attributed to the same country.

Let's begin by looking at the unique values of the **Country** column:

```python
df.Country.value_counts()
```

Let's go through each value and rename them so they are unified.

```python
df['Country'] = df['Country'].replace({"SG": "Singapore", "unknown": "Unknown", "india": "India", "INDIA": "India", "srilanka": "Sri Lanka",
                                       "Italia": "Italy", "UK": "United Kingdom", "jamaica": "Jamaica", "BARBADOS": "Barbados", "MY": "Malaysia"})
```

```python
df.Country.value_counts()
```
