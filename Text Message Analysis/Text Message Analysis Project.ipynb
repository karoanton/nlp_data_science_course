### Noise removal

In order to clean the text of the *Message* column in the dataframe, I want to remove the following 'noise' from the data:
- Remove punctuation and special characters
- Make all text lowercase

First, let's go ahead and remove the punctuation and special characters.

```python
df['Message'] = df['Message'].str.replace(r'[^0-9a-zA-Z ]', '', regex=True)
```

```python
df.head()
```

Now I will edit the text to make it all lowercase.

```python
df['Message'] = df['Message'].str.lower()
```

```python
df.head()
```

### Tokenization

I now want to tokenize the messages included in *Message*. Therefore, I will import the NLTK library, including the functions **word_tokenize()** and **sent_tokenize** to tokenize both the words and sentences in each text message.

```python
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
df['Word Tokenized'] = df.apply(lambda row: nltk.word_tokenize(row['Message']), axis=1)
```

```python
df['Sentence Tokenized'] = df.apply(lambda row: nltk.sent_tokenize(row['Message']), axis=1)
```

### PoS Tagging

```python
from nltk.corpus import wordnet
from collections import Counter
```

```python
def get_part_of_speech(word):
  probable_part_of_speech = wordnet.synsets(word)
  
  pos_counts = Counter()

  pos_counts["n"] = len(  [ item for item in probable_part_of_speech if item.pos()=="n"]  )
  pos_counts["v"] = len(  [ item for item in probable_part_of_speech if item.pos()=="v"]  )
  pos_counts["a"] = len(  [ item for item in probable_part_of_speech if item.pos()=="a"]  )
  pos_counts["r"] = len(  [ item for item in probable_part_of_speech if item.pos()=="r"]  )
  
  most_likely_part_of_speech = pos_counts.most_common(1)[0][0]
  return most_likely_part_of_speech
```

```python
df['Lemmatized and POS Tagged'] = df['Word Tokenized, Stemmed, and Lemmatized'].apply(lambda words: [lemmatizer.lemmatize(word, get_part_of_speech(word)) for word in words])
```
